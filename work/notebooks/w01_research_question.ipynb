## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The dataset contains 30,000 content items and 44 columns. Pages trending up have a higher average content age (288.48 days) than pages trending down (236.18 days). Their mean CTR is also higher (0.566 versus 0.324). These observed differences make content performance worth investigating further, but they do not show that content age causes better performance. The dataset documentation defines CTR values as percentages multiplied by 100, so these values should not be read as whole-number percentages.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/mzaheer1070/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

print("Dataset shape:", df.shape)
print("\nAverage content age by trend direction:")
print(
    df.groupby("trend_direction")["content_age_days"]
      .agg(["count", "mean", "median"])
      .round(2)
)

print("\nAverage CTR by trend direction:")
print(df.groupby("trend_direction")["ctr"].mean().round(3))
```
